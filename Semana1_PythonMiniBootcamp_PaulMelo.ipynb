# Retos - Validación visual

```python
###   CALCULADORA DE PROPINAS   ###

# ------------------------------- #
## ---       Encabezado      --- ##
# ------------------------------- #

nombre   = input("¿Cúal es su nombre? ")
cliente  = input("¿Cúal es el nombre del cliente? ")
moneda   = input("¿En qué moneda? (MXN/USD/EUR): ").upper()
impuesto = input("¿Qué porcentaje de propina? (15, 18, 20): ")

# ------------------------------- #
## ---    Datos de compra    --- ##
# ------------------------------- #
cuenta           = float (input("¿Valor de la cuenta? ($): "))
impuesto_comsumo = int   (impuesto)
cant_personas    = int   (input("¿Cuantas personas?: "))

# ------------------------------- #
## ---      Calculadora      --- ##
# ------------------------------- #
propina = cuenta * (impuesto_comsumo / 100)
total = cuenta + propina
por_persona = total / cant_personas

# ---------------------------------- #
# - Condicionales IF - ELIF - ELSE - #
# ---------------------------------- #

message_imp = "Propina estándar" if impuesto_comsumo == 15 else ("Buena propina" if impuesto_comsumo == 18 else "Propina excelente")

recibo = f"""
=========================================
            TICKET DE COMPRA
=========================================
Cliente             : {cliente.upper()}
-----------------------------------------

Cuenta             $: {moneda} - {cuenta:.2f}
Impuesto consumo (%): {impuesto_comsumo}  - {message_imp}

=========================================
            RESUMEN {nombre.upper()}
=========================================

Cuenta             $: {moneda} - {cuenta:.2f}
Propina            $: {moneda} - {propina:.2f}
Total              $: {moneda} - {total:.2f}


-----------------------------------------

Personas            : {cant_personas}
Cuenta x persona   $: {moneda} - {por_persona:.2f}
=========================================
"""
print(recibo)
```
