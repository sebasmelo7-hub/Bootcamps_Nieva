```python
###              AGENDA CONTACTOS COMPLETA              ###

# El programa entero es:

# Definir la lista contactos, el set emails_registrados y la tupla OPCIONES_MENU.
# Entrar a un ciclo while True que muestra el menú.
# Leer la opción del usuario con input().
# Ramificar con if/elif/else para ejecutar la operación elegida.
# Salir del ciclo con break cuando elija "Salir".

contactos          = []
emails_registrados = set()
OPCIONES_MENU = ("Agregar", "Listar", "Buscar", "Eliminar", "Salir")

while True:
    # Menú principal
    print("\n" + "=" * 40)
    print("***     AGENDA DE CONTACTOS     ***")
    print("=" * 40)
    for i, opcion in enumerate(OPCIONES_MENU, start=1):
        print(f"{i}. {opcion}")
    print("=" * 40)


    opcion = input("Elige una opcion (1-5): ")
   
       # Usamos if/else/elif para decidir que accion realizar con la variable - opcion
       # Opcion 1: Agregar
    if opcion == "1":
        print("\n----     AGREGAR CONTACTO     ---")
        nombre   = input("Nombre: ")
        telefono = input("Telefono: ")
        email    = input("Email: ")

        if email in emails_registrados:
            print(f"Error: el email '{nuevo['email']}' ya existe en la agenda.")
        else:
            nuevo = {"nombre": nombre, "telefono": telefono, "email": email}
            contactos.append(nuevo)
            emails_registrados.add(email)
            print(f"Contacto '{nombre}' agregado correctamente.")
            print(f"Total: {len(contactos)} contacto(s)")

    # Opcion 2: Listar
    elif opcion == "2":
        print("\n---     LISTA DE CONTACTOS     --")
        if len(contactos) == 0:
            print("No existen contactos registrados")
        else:
            for i, contacto in enumerate(contactos, start=1):
                print(f"\n   {i}. {contacto['nombre']}")
                print(f"     Tel: {contacto['telefono']}")
                print(f"   Email: {contacto['email']}")
            print(f"\nTotal {len(contactos)} contactos(s)")
        
    # Opcion 3: Buscar
    elif opcion == "3":
        print("\n----     BUSCAR CONTACTO     ----")
        if len(contactos) == 0:
            print("No existen contactos registrados. Agenda vacía")
        else:
            busqueda = input("Escribe el nombre a buscar: ")

            encontrados = []
            for contacto in contactos:
                if busqueda.lower() in contacto["nombre"].lower():
                    encontrados.append(contacto)

            if len(encontrados) == 0:
                print(f"No se encontraron contactos con '{busqueda}'.")
            else:
                print(f"\nSe encontraron {len(encontrados)} resultado(s):")
                for i, c in enumerate(encontrados, start=1):
                    print(f"  {i}. {c['nombre']} | {c['telefono']} | {c['email']}")

    # Opcion 4: Eliminar
    elif opcion == "4":        
        print("\n---     ELIMINAR CONTACTO     ---")
        if len(contactos) == 0:
                    print("No existen contactos para eliminar")
        else:
            nombre_eliminar = input("Nombre del contacto a eliminar: ")

            contacto_encontrado = None
            for contacto in contactos:
                if nombre_eliminar.lower() in contacto["nombre"].lower():
                    contacto_encontrado = contacto
                    break

            if contacto_encontrado is None:
                print(f"No se encontró ningún contacto con el nombre '{nombre_eliminar}'.")
            else:
                print(f"\nContacto encontrado:")
                print(f"Nombre:   {contacto_encontrado['nombre']}")
                print(f"Teléfono: {contacto_encontrado['telefono']}")
                print(f"Email:    {contacto_encontrado['email']}")

                confirmacion = input(f"  ¿Eliminar '{contacto_encontrado['nombre']}'? (si/no): ").lower()
                if confirmacion.lower() == "si":
                    emails_registrados.discard(contacto_encontrado["email"])
                    contactos.remove(contacto_encontrado)
                    print(f"Contacto '{contacto_encontrado['nombre']}' eliminado.")
                    print(f"Contactos restantes: {len(contactos)}")
                else:
                    print("Eliminacion de contacto cancelada.")

    # Opción 5: Salir
    elif opcion == "5":
        print("\nHasta luego. Tu agenda se cerrará.")
        break
    
    # Opción inválida
    else:
        print("Opción no válida. Elige un número del 1 al 5.")  

```

```python
###              RETO B - LISTAR ORDENADO ALFABETICAMENTE              ###

# Modifica la opción "Listar" para mostrar los contactos ordenados por nombre. 
# Usa sorted() con el parámetro key

contactos          = []
emails_registrados = set()
OPCIONES_MENU = ("Agregar", "Listar", "Buscar", "Eliminar", "Editar", "Salir")

while True:
    # Menú principal
    print("\n" + "=" * 40)
    print("***     AGENDA DE CONTACTOS     ***")
    print("=" * 40)
    for i, opcion in enumerate(OPCIONES_MENU, start=1):
        print(f"{i}. {opcion}")
    print("=" * 40)


    opcion = input("Elige una opcion (1-6): ")
   
       # Usamos if/else/elif para decidir que accion realizar con la variable - opcion
       # Opcion 1: Agregar
    if opcion == "1":
        print("\n----     AGREGAR CONTACTO     ---")
        nombre   = input("Nombre: ")
        telefono = input("Telefono: ")
        email    = input("Email: ")

        if email in emails_registrados:
            print(f"Error: el email '{nuevo['email']}' ya existe en la agenda.")
        else:
            nuevo = {"nombre": nombre, "telefono": telefono, "email": email}
            contactos.append(nuevo)
            emails_registrados.add(email)
            print(f"Contacto '{nombre}' agregado correctamente.")
            print(f"Total: {len(contactos)} contacto(s)")

    # Opcion 2: Listar
    elif opcion == "2":
        print("\n---     LISTA DE CONTACTOS     --")
        if len(contactos) == 0:
            print("No existen contactos registrados")
        else:
            contactos_ordenados = sorted(contactos, key=lambda c:c['nombre'].lower())
            for i, contacto in enumerate(contactos_ordenados, start=1):
                print(f"\n   {i}. {contacto['nombre']}")
                print(f"     Tel: {contacto['telefono']}")
                print(f"   Email: {contacto['email']}")
            print(f"\nTotal {len(contactos)} contactos(s)")
        
    # Opcion 3: Buscar
    elif opcion == "3":
        print("\n----     BUSCAR CONTACTO     ----")
        if len(contactos) == 0:
            print("No existen contactos registrados. Agenda vacía")
        else:
            busqueda = input("Escribe el nombre a buscar: ")

            encontrados = []
            for contacto in contactos:
                if busqueda.lower() in contacto["nombre"].lower():
                    encontrados.append(contacto)

            if len(encontrados) == 0:
                print(f"No se encontraron contactos con '{busqueda}'.")
            else:
                print(f"\nSe encontraron {len(encontrados)} resultado(s):")
                for i, c in enumerate(encontrados, start=1):
                    print(f"  {i}. {c['nombre']} | {c['telefono']} | {c['email']}")

    # Opcion 4: Eliminar
    elif opcion == "4":        
        print("\n---     ELIMINAR CONTACTO     ---")
        if len(contactos) == 0:
                    print("No existen contactos para eliminar")
        else:
            nombre_eliminar = input("Nombre del contacto a eliminar: ")

            contacto_encontrado = None
            for contacto in contactos:
                if nombre_eliminar.lower() in contacto["nombre"].lower():
                    contacto_encontrado = contacto
                    break

            if contacto_encontrado is None:
                print(f"No se encontró ningún contacto con el nombre '{nombre_eliminar}'.")
            else:
                print(f"\nContacto encontrado:")
                print(f"Nombre:   {contacto_encontrado['nombre']}")
                print(f"Teléfono: {contacto_encontrado['telefono']}")
                print(f"Email:    {contacto_encontrado['email']}")

                confirmacion = input(f"  ¿Eliminar '{contacto_encontrado['nombre']}'? (si/no): ").lower()
                if confirmacion.lower() == "si":
                    emails_registrados.discard(contacto_encontrado["email"])
                    contactos.remove(contacto_encontrado)
                    print(f"Contacto '{contacto_encontrado['nombre']}' eliminado.")
                    print(f"Contactos restantes: {len(contactos)}")
                else:
                    print("Eliminacion de contacto cancelada.")


    # Opción 5: Editar
    elif opcion == "5":
        print("\n----     EDITAR CONTACTO     ----")
        if len(contactos) == 0:
            print("No hay contactos para editar")
        else:
            editar_contacto = input("Nombre del contacto a editar: ")

            contacto_encontrado = None
            for contacto in contactos:
                if editar_contacto.lower() in contacto["nombre"].lower():
                    contacto_encontrado = contacto
                    break

            if contacto_encontrado is None:
                print(f"No se encontró ningún contacto con el nombre '{nombre_eliminar}'.")
            else:
                print(f"\nContacto encontrado:")
                print(f"Nombre:   {contacto_encontrado['nombre']}")
                print(f"Teléfono: {contacto_encontrado['telefono']}")
                print(f"Email:    {contacto_encontrado['email']}")

                confirmacion = input(f"  ¿Editar'{contacto_encontrado['nombre']}'? (si/no): ").lower()
                if confirmacion.lower() == "si":
                    telefono = input("Telefono: ")
                    email    = input("Email: ")
            
                    if email in emails_registrados:
                        print(f"Error: el email '{nuevo['email']}' ya existe en la agenda.")
                    else:
                        contacto_encontrado['telefono'] = telefono
                        contacto_encontrado['email']    = email
                        print(f"Contacto editado correctamente. {contacto_encontrado['nombre']}|{telefono}|{email}")
                else:
                    print("Edicion de contacto cancelada.")


    # Opción 6: Salir
    elif opcion == "6":
        print("\nHasta luego. Tu agenda se cerrará.")
        break
    
    # Opción inválida
    else:
        print("Opción no válida. Elige un número del 1 al 6.")  

```

```python
###              AGENDA CONTACTOS - 2              ###

# Menú principal

contactos = []

# Usamos while True  para que el ciclo no termine hasta qye elijan - opcion  5. Salir
while True:
    # Menú principal
    print("=" * 35)
    print("***     AGENDA DE CONTACTOS     ***")
    print("=" * 35)
    print("1. Agregar contacto")
    print("2. Ver todos los contactos")
    print("3. Buscar contacto")
    print("4. Eliminar contacto")
    print("5. Salir")
    print("=" * 35)

    opcion = input("Elige una opcion (1-5): ").strip()
    # Usamos .strip() para eliminar espacio en blanco al incio y final

    # Usamos if/else/elif para decidir que accion realizar con la variable - opcion
    # Opcion 1: Agregar
    if opcion == "1":
        print("\n----     AGREGAR CONTACTO     ---")
        nombre   = input("Nombre: ").strip()
        telefono = input("Telefono: ").strip()
        email    = input("Email: ").strip()

        if not nombre:
            print("El nombre no puede estar vacío.")
        else:
            nuevo = {"nombre": nombre, "telefono": telefono, "email": email}
            contactos.append(nuevo)
            print(f"Contacto '{nombre}' agregado.")

    # Opcion 2: Listar
    elif opcion == "2":
        print("\n---     LISTA DE CONTACTOS     --")
        if len(contactos) == 0:
            print("No existen contactos registrados")
        else:
            for i, contacto in enumerate(contactos, start=1):
                print(f"\n   {i}. {contacto['nombre']}")
                print(f"     Tel: {contacto['telefono']}")
                print(f"   Email: {contacto['email']}")
            print(f"\nTotal {len(contactos)} contactos(s)")

    # Opcion 3: Buscar
    elif opcion == "3":
        print("\n----     BUSCAR CONTACTO     ----")
        busqueda   = input("Ingresa el contacto a buscar (nombre): ").strip
        encontrado = False

        for contacto in contactos:
            # .lower() en ambos lados para búsqueda sin distinción de mayúsculas
            if busqueda.lower() in contacto["nombre"].lower():
                print(f"\nContacto encontrado:")
                print(f"  Nombre:   {contacto['nombre']}")
                print(f"  Teléfono: {contacto['telefono']}")
                print(f"  Email:    {contacto['email']}")
                encontrado = True
                break # Cuando encuentra el nombre buscado, se detiene.

        if not encontrado:
            print(f"No se encontró ningún contacto con nombre: '{busqueda}'.")
        

    # Opcion 4: Eliminar
    elif opcion == "4":        
        print("\n---     ELIMINAR CONTACTO     ---")
        if len(contactos) == 0:
                    print("No existen contactos para eliminar")
        else:
            nombre_eliminar = input("Nombre del contacto a eliminar: ")
            indice          = -1

            for i, contacto in enumerate(contactos):
                if nombre_eliminar.lower() == contacto["nombre"].lower():
                    indice = i
                    break

            if indice >= 0:
                contacto_eliminado = contactos.pop(indice)
                print(f"Contacto '{contacto_eliminado['nombre']}' eliminado.")
                print(f"Contactos restantes: {len(contactos)}")
            else:
                print(f"No se encontró ningún contacto con el nombre '{nombre_eliminar}'.")

    # Opción 5: Salir
    elif opcion == "5":
        print("\nHasta luego. Tu agenda se cierra.")
        break
    
    # Opción inválida
    else:
        print("Opción no válida. Elige entre 1 y 5.")   
```
