Python tiene cuatro tipos de colecciones principales. Cada una tiene características distintas que la hacen ideal para diferentes situaciones.
##    LISTAS - list    #
1. Usa corchetes []
2. Puede tener elementos repetidos
3. Mantiene el orden en que agregaste los elementos
4. Puedes cambiar sus elementos (agregar, eliminar, modificar)

Usa una lista cuando:
• Tienes una secuencia de elementos del mismo tipo
• El orden importa
• Necesitas agregar o quitar elementos
• Ejemplos: lista de tareas, historial de compras, nombres de estudiantes

##    DICCIONARIO - dict    #
1. Usa llaves {} con pares clave: valor
2. Accedes a los datos por nombre (clave), no por posición
3. No permite claves repetidas
4. Puedes agregar, modificar y eliminar pares

Usa un diccionario cuando:
• Tienes datos con nombre (atributos)
• Necesitas acceder a la información por su nombre, no por su posición
Ejemplos: datos de un usuario, configuración de una app, un contacto de agenda

##    CONJUNTO - set    #
1. Usa llaves {} pero sin pares clave: valor
2. Elimina automáticamente los duplicados
3. No tiene un orden definido (el orden puede variar)
4. Ideal para operaciones matemáticas de conjuntos

Usa un conjunto cuando:
• Necesitas que no haya duplicados
• Vas a hacer operaciones matemáticas de conjuntos (intersección, unión, diferencia)
• Ejemplos: lista de emails únicos, palabras de un texto sin repetir

##    TUPLA - tuple    #
1. Usa paréntesis ()
2. Una vez creada, no puedes modificarla (es inmutable)
3. Mantiene el orden
4. Útil para datos que no deben cambiar

Usa una tupla cuando:
• Los datos no deben cambiar durante el programa
• Necesitas una clave compuesta en un diccionario
• Una función necesita devolver múltiples valores
Ejemplos: coordenadas GPS, fecha de nacimiento, RGB de un color

##    Slicing: extraer porciones de una lista    #
El slicing te permite extraer una parte de la lista. La sintaxis es lista[start:stop:step].

1. start — índice donde empieza (incluido)
2. stop — índice donde termina (excluido — el elemento en stop NO se incluye)
3. step — paso (cuántos saltar entre elementos, opcional)
Punto clave: el slicing devuelve una nueva lista — no modifica la original.

##    Métodos principales de las listas    #
Los métodos son acciones que puedes hacer con una lista. Se usan con la sintaxis lista.metodo()

1. append(elemento) — agrega un elemento al final
2. insert(indice, elemento) — agrega un elemento en una posición específica
3. remove(elemento) — elimina la primera aparición del elemento especificado
4. pop(indice) — elimina y devuelve el elemento en el índice dado. Sin argumento, elimina el último:
5. clear() — elimina todos los elementos
6. sort() — ordena la lista en su lugar (modifica la lista original). Usa sort(reverse=True) para realizar orden descendente
7. reverse() — invierte el orden de la lista (modifica la original)
8. index(elemento) — devuelve el índice de la primera aparición del elemento. Usa index(elemento, indice) para Buscar desde una posición específica
9. count(elemento) — cuenta cuántas veces aparece el elemento
10. Para cambiar un elemento, accedes por índice y asignas el nuevo valor

### CONJUNTOS (set) ###

##    CONJUNTO - set    #
1. Usa llaves {} pero sin pares clave: valor
2. Crear conjunto con set(). Para crear un set vacío usa set().
3. Agregar elementos con add(elemento)
4. Eliminas elementos con remove(elemento) — eliminar (lanza error si no existe). remove() cuando estás seguro de que el elemento existe
5. Eliminar elementos con discard(elemento) — eliminar sin error si no existe. discard() cuando no sabes si existe y no quieres que falle

# Propiedades clave de los sets
1. Sin duplicados — cada elemento aparece exactamente una vez
2. Sin orden — no puedes acceder por índice (set[0] da error)
3. Mutable — puedes agregar y eliminar elementos

# OPERACIONES DE CONJUNTOS
1. Unión — todos los elementos de ambos --> método union(), Con operador |
2. Intersección — elementos que están en AMBOS --> método intersection(), Con operador &
3. Diferencia — elementos en a que NO están en b --> método difference(), Con operador -, La diferencia no es simétrica: a - b ≠ b - a
---------------------------------------------------------------------------------------------------------------------------------
4. El patrón list(set(lista)) es una forma idiomática en Python para eliminar duplicados de una lista.
5. sorted() — para ordenar la lista (modifica la lista original)
6. El operador in en un set es mucho más rápido que en una lista, especialmente cuando tienes miles de elementos. Para una agenda pequeña no importa, pero es un buen hábito.
7. El patrón sorted(list(set(lista))) es muy común: primero elimina duplicados con set(), luego convierte a lista con list(), luego ordena con sorted(). Se hace de adentro hacia afuera.

### TABLA DE DECISION ###
Pregunta clave                                            - Respuesta --> Colección
¿Tienen los datos un nombre para cada campo?  (atributos) -	Sí        --> Diccionario {}
¿Necesito que no haya duplicados y el orden no importa?   - Sí        --> Conjunto set()
¿Los datos NO deben cambiar nunca?                        - Sí        --> Tupla ()
¿Nada de lo anterior aplica?                              - —         --> Lista []

## ARBOL DE DECISION ##
¿Los datos tienen campos con nombre?
    ├── Sí → Diccionario {"nombre": valor, ...}
    └── No
        ├── ¿Los datos no pueden repetirse?
        │   ├── Sí → Set {valor1, valor2, ...}
        │   └── No
        │       ├── ¿Los datos no deben cambiar?
        │       │   ├── Sí → Tupla (valor1, valor2, ...)
        │       │   └── No → Lista [valor1, valor2, ...]
-----------------------------------------------------------------------------------------------------------------------------------
La clave de la decisión: si tiene campos con nombre → diccionario. Si sin duplicados → set. Si no cambia → tupla. Si ninguno de los anteriores (o si necesitas orden + duplicados + mutabilidad) → lista.

## FUNCIONES ##
1. len() — contar elementos. Devuelve el número de elementos de cualquier colección
2. sum(), min(), max() — estadísticas básicas. También funcionan con strings (min y max usan orden alfabético)
3. .sort() — modifica la lista original. Usa .sort() cuando no necesitas el orden original
4. sorted() — devuelve una lista nueva, la original no cambia. Usa sorted() cuando necesitas mantener el original
5. enumerate() — iterar con índice. Cuando iteras una lista con for, normalmente solo tienes el elemento. Con enumerate() obtienes el índice también
6. zip() — combinar dos listas en pares. zip() toma dos (o más) listas y las combina elemento por elemento
7. reversed() — iterar al revés sin modificar. reversed() te permite recorrer una secuencia de atrás hacia adelante sin modificar la original
    71. Diferencia con [::-1]: El slicing crea una nueva lista en memoria. reversed() es un iterador — más eficiente en memoria para listas grandes, pero el resultado es el mismo para uso práctico.
8. Operador in y not in — verificar membresía. El operador in es especialmente eficiente con sets — lo verifica en tiempo constante sin importar el tamaño del set

```python
###              MEJORA 3 - TELEFONO CONTACTOS SOLO NÚMERO              ###

def es_telefono_valido(telefono):
    """Valida que el teléfono solo contenga dígitos, guiones y espacios."""
    if not telefono:
        return True  # Teléfono vacío es aceptable

    caracteres_validos = set("0123456789-+ ()")
    for caracter in telefono:
        if caracter not in caracteres_validos:
            return False
    return True

    # En agregar_contacto():
    telefono = input("  Teléfono: ").strip()
    if telefono and not es_telefono_valido(telefono):
        print("  Teléfono inválido — solo dígitos, guiones y espacios.")
        return
```

```python
###              CODIGO COMPLETO Y FUNCIONAL              ###

# ============================================
# AGENDA DE CONTACTOS
# Mini-Proyecto — Week 3: Colecciones
# Python Mini-Bootcamp
# ============================================

# Lista contactos
contactos = []

# Función Menú Principal
def menu_principal():
    """Muestra el menú principal de la agenta"""
    # Menú principal
    print("=" * 35)
    print("***     AGENDA DE CONTACTOS     ***")
    print("=" * 35)
    print("1. Agregar contacto")
    print("2. Ver todos los contactos")
    print("3. Buscar contacto")
    print("4. Eliminar contacto")
    print("5. Salir")
    print("=" * 35)

# Función Agregar Contacto
def agregar_contacto():
    """Pide los datos al usuario y agrega el contacto a la lista."""
    print("\n----     AGREGAR CONTACTO     ---")
    nombre = input("Nombre: ").strip()

    if not nombre:
       print("El nombre no puede estar vacío.")
       return

    telefono = input("Telefono: ").strip()
    email    = input("Email: ").strip()

    nuevo_contacto = {
    "nombre": nombre,
    "telefono": telefono,
    "email": email
    }

    contactos.append(nuevo_contacto)
    print(f"\nContacto '{nombre}' agregado exitosamente.")
    print(f"Total de contactos: {len(contactos)}")
    print("Datos guardados:", nuevo_contacto)

# Función Listar Contactos
def listar_contactos():
    """Muestra todos los contactos numerados."""
    print("\n---     LISTA DE CONTACTOS     --")
    if len(contactos) == 0:
        print("No existen contactos en la agenda")
        print("Usa la opción 1 para agregar tu primer contacto.")
        return

    for i, contacto in enumerate(contactos, start=1):
        print(f"\n   {i}. {contacto['nombre']}")
        print(f"     Tel: {contacto['telefono']}")
        print(f"   Email: {contacto['email']}")
    print(f"\nTotal {len(contactos)} contactos(s)")

# Función Buscar Contacto
def buscar_contacto():
    """Busca un contacto por nombre (búsqueda parcial, sin distinción de mayúsculas)."""
    print("\n----     BUSCAR CONTACTO     ----")
    if len(contactos) == 0:
        print("No hay contactos en la agenda")
        return
    
    busqueda   = input("Ingresa el contacto a buscar (nombre): ").strip()
        
    if not busqueda:
        print("El nombre no puede estar vacío. Escribe un nombre para buscar")
        return

    encontrado = False
    
    for contacto in contactos:
        # .lower() en ambos lados para búsqueda sin distinción de mayúsculas
        if busqueda.lower() in contacto["nombre"].lower():
            print(f"\nContacto encontrado:")
            print(f"  Nombre:   {contacto['nombre']}")
            print(f"  Teléfono: {contacto['telefono']}")
            print(f"  Email:    {contacto['email']}")
            encontrado = True
            break # Cuando encuentra el nombre buscado, se detiene.
    
    if not encontrado:
        print(f"No se encontró ningún contacto con nombre: '{busqueda}'.")

# Función Eliminar Contacto
def eliminar_contacto():
    """Busca un contacto por nombre exacto y lo elimina."""
    print("\n---     ELIMINAR CONTACTO     ---")
    if len(contactos) == 0:
        print("No existen contactos para eliminar")
        return

    # Mostramos la lista para que el usuario sepa que contactos tiene
    listar_contactos()

    nombre_eliminar = input("Nombre del contacto a eliminar: ")
    indice          = -1

    for i, contacto in enumerate(contactos):
        if nombre_eliminar.lower() == contacto["nombre"].lower():
            indice = i
            break

    if indice >= 0:
        confirmacion = input(f"  ¿Eliminar '{contactos[indice]['nombre']}'? (s/n): ").strip().lower()
        if confirmacion == "s":
            contacto_eliminado = contactos.pop(indice)
            print(f"Contacto '{contacto_eliminado['nombre']}' eliminado.")
            print(f"Contactos restantes: {len(contactos)}")
        else:
            print("Eliminacion de contacto cancelada.")

    else:
        print(f"No se encontró ningún contacto con el nombre '{nombre_eliminar}'.")


# ============================================
# PROGRAMA PRINCIPAL
# ============================================
print("Bienvenido a tu Agenda de Contactos")

while True:
    menu_principal()
    opcion = input("  Elige una opción (1-5): ").strip()

    if opcion == "1":
        agregar_contacto()
    elif opcion == "2":
        listar_contactos()
    elif opcion == "3":
        buscar_contacto()
    elif opcion == "4":
        eliminar_contacto()
    elif opcion == "5":
        print("\nHasta luego. ¡Sigue programando!")
        break
    else:
        print("\n  Opción no válida. Elige un número del 1 al 5.")
```

```python
###   Troubleshooting 6 - FUNCIONES INTEGRADAS   ###

# Problema 1: sum() con lista de strings #
# Usa sum() solo con listas numéricas. Para concatenar strings, usa "".join(lista) o un loop.
nombres = ["Ana", "Carlos", "María"]
concatenado = ", ".join(nombres)
print(concatenado) # Output: Ana, Carlos, María

print("-" * 40)

# Problema 2: Olvidar que sorted() devuelve lista, no el tipo original #
# sorted() siempre devuelve una lista, sin importar el tipo de entrada.
mi_tupla = (3, 1, 4, 1, 5)
resultado = tuple(sorted(mi_tupla))
print(type(resultado)) # Output: <class 'tuple'>
print(resultado) # Output: (1, 1, 3, 4, 5)
print("-" * 40)

# Problema 3: enumerate() sin desempacar correctamente #
frutas = ["manzana", "banana", "naranja"]

for i, fruta in enumerate(frutas):
    print(f"{i+1}: {fruta}")
print("-" * 40)

# Problema 4: zip() con listas de diferente longitud (comportamiento silencioso) #
# Verifica que las listas tienen la misma longitud antes de usar zip().
nombres = ["Ana", "Carlos", "María", "Luis"]
notas = [9.5, 8.0]

if len(nombres) != len(notas):
    print(f"Advertencia: las listas tienen diferente longitud ({len(nombres)} vs {len(notas)})")
else:
    for nombre, nota in zip(nombres, notas):
        print(nombre, nota) # Output: Advertencia: las listas tienen diferente longitud (4 vs 2)
print("-" * 40)

# Problema 5: max() y min() con lista vacía #
# Verifica que la lista no está vacía antes de llamar max() o min()
vacia = []

if vacia:
    print(max(vacia))
else:
    print("La lista está vacía") # Output: La lista está vacía
```

```python
###   Troubleshooting 5 - CONJUNTOS (set)   ###

# Problema 1: TypeError al intentar modificar una tupla #
# Si necesitas modificar los datos, usa una lista. Si definitivamente son datos fijos, déjalos como tupla

# Problema 2: Crear una tupla de un elemento olvidando la coma #
# Los paréntesis solos son paréntesis matemáticos, no crean una tupla --> Agrega una coma después del elemento.
es_tupla = (42,)
print(type(es_tupla)) # Output: <class 'tuple'>
print(es_tupla.count(42)) # Output: 1

# Problema 3: Confundir desempaquetado con número incorrecto de variables #
# La tupla tiene 3 elementos pero solo defines 2 variables para desempacar. --> El número de variables debe coincidir con el número de elementos
punto = (10, 20, 30)
x, y, z = punto
print(x, y, z) # Output: 10 20 30

# O usa * para capturar el resto en una lista
x, *resto = punto
print(x)      # Output: 10
print(resto)  # Output: [20, 30]
```

```python
### CONJUNTOS (set) ###

## OPERACIONES CON CONJUNTOS ##
# Unión — todos los elementos de ambos #
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# Con método union()
union = a.union(b)
print(union) # Output: {1, 2, 3, 4, 5, 6}
print("-" * 40)

# Con operador |
print(a | b) # Output: {1, 2, 3, 4, 5, 6}

print("=" * 40)
# Intersección — elementos que están en AMBOS #
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# Con método intersection()
interseccion = a.intersection(b)
print(interseccion) # Output: {3, 4}
print("-" * 40)

# Con operador &
print(a & b)# Output: {3, 4}

print("=" * 40)
# Diferencia — elementos en a que NO están en b #
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# Con método difference()
diferencia = a.difference(b)
print(diferencia) # Output: {1, 2}
print("-" * 40)

# Con operador -
print(a - b) # Output: {1, 2}
print("-" * 40)
# La diferencia no es simétrica: a - b ≠ b - a
print(b - a) # Output: {5, 6}
```

```python
###   DICCIONARIOS Y METODOS - EJERCICIO 2 - ACCESO CON get()    ###

# Dado el diccionario producto = {"nombre": "Laptop", "precio": 15999.99, "marca": "TechBrand"}

# Imprime el nombre del producto
# Intenta obtener el campo "stock" — si no existe, imprime "Stock no disponible"
# Intenta obtener el campo "descuento" — si no existe, imprime 0

producto = {"nombre": "Laptop", "precio": 15999.99, "marca": "TechBrand"}

# 1. Nombre del producto
print(producto["nombre"])

# 2. Campo "stock"
print(producto.get("stock", "Stock no disponible"))

if "stock" in producto:
    print(producto["stock"])
else:
    print("Stock no disponible")


# 2. Campo "descuento"
print(producto.get("descuento", 0))

if "descuento" in producto:
    print(producto["descuento"])
else:
    print(0)
```

```python
###   LISTAS Y METODOS - EJERCICIO 4 - BUSCAR y CONTAR    ###

# Dada la lista calificaciones = [8, 9, 7, 10, 8, 6, 8, 9, 10, 7]:

# Cuenta cuántas veces aparece el 8
# Encuentra en qué índice está el primer 10
# Verifica si el 5 está en la lista (usa in)
# Imprime todas las calificaciones que son mayores a 8

calificaciones = [8, 9, 7, 10, 8, 6, 8, 9, 10, 7]

# 1. ¿Cántas veces aparece el 8?
# count(elemento) — cuenta cuántas veces aparece el elemento.
ocho = calificaciones.count(8)
print(f"El 8 aparece {ocho} veces")
print("-" * 40)

# 2. Indice primer 10
# index(elemento) — devuelve el índice de la primera aparición del elemento.
Ind_diez = calificaciones.index(10)
print(f"Indice primer 10: ", Ind_diez)
print("-" * 40)

# 3. ¿Existe el 5 en la lista?
if 5 in calificaciones:
    print("Encontre al 5")
else:
    print("El 5 no se encuentra en la lista")
    print("-" * 40)

# 4. Calificaciones > 8
for calificacion in calificaciones:
    if calificacion > 8:
        print(calificacion)
    else:
        continue

```

```python
###   LISTAS Y METODOS - EJERCICIO 3 - SLICING    ###

##    Slicing: permite extraer una parte de la lista. La sintaxis es lista[start:stop:step].

# Dada la lista letras = ["a", "b", "c", "d", "e", "f", "g", "h"], usa slicing para obtener:

# Las primeras 3 letras
# Las últimas 3 letras
# Las letras de la posición 2 a la 5 (sin incluir la 5)
# Las letras en posiciones pares (0, 2, 4, 6)
# La lista al revés

letras = ["a", "b", "c", "d", "e", "f", "g", "h"]

# 1. Las primeras 3 letras
print (letras[0:3])
print("-" * 40)

# 2. Las últimas 3 letras
print (letras[-3:])
print("-" * 40)

# 3. Letras de la posición 2 a la 5 (sin incluir la 5)
print (letras[2:5])
print("-" * 40)

# 4. Las letras en posiciones pares (0, 2, 4, 6)
print (letras[::2])
print("-" * 40)

# 4. Lista al revés
print (letras[::-1])
```

```python
###   LISTAS Y METODOS - EJERCICIO 2 - OPERACIONES DE LISTA    ###

# Dada la lista numeros = [5, 2, 8, 1, 9, 3, 7]:

# Ordénala de menor a mayor (usando sort())
# Imprime el número más pequeño y el más grande
# Invierte el orden con reverse()
# Elimina el primer elemento con pop()

numeros = [5, 2, 8, 1, 9, 3, 7]

# 1. Orderar
numeros.sort()
print (numeros)
print("-" * 40)

# 2. Número más pequeño y más grande
print("No. más pequeño:", numeros[0])
print("No. más grande :", numeros[-1])
print("-" * 40)

# 3. Invertir orden
numeros.reverse()
print (numeros)
print("-" * 40)

# 4. Elimina el primer elemento
first = numeros[0]
print(f"Indice 0 de la lista, eliminado: {first}")
numeros.pop(0)
print (f"Lista final {numeros}")
print("-" * 40)
```
