```python
###   CONTINUE  JUEGO ADIVINA EL NÚMERO   ###

num_secret = 40
intentos = 0
max_intentos = 7

# Usamos ciclo while
while intentos < max_intentos:
    entrada = input(f"Intento {intentos + 1}/{max_intentos} — Adivina el número (1-100): ")

    # Validación 2: ¿es número?
    if not entrada.isnumeric():
      print("Eso no es un número. Ingresa un número.")
      continue

    # 2. Si pasó la validación, ahora sí lo convertimos a entero
    num_adivina = int(entrada)
    intentos += 1

    print("=" * 40)
    if 1 <= num_adivina <= 100:
      if num_adivina > num_secret:
          print("Muy alto")
      elif num_adivina < num_secret:
          print("Muy bajo")
      else:
          print(f"¡Correcto! Lo lograste en {intentos} intentos")
          break
    else:
      print("El numero debe estar en un rango de 1 a 100")
    print("-" * 40)

else:
  print(f"Se acabaron los intentos. El número era {num_secret}.")
```

```python
###   JUEGO ADIVINA EL NÚMERO   ###

import random

print("=" * 40)
print("        JUEGO ADIVINA EL NÚMERO")
print("=" * 40)

num_secret = random.randint(1, 100)
max_intentos = 7


while True:
  intentos = 0
  print(f"\nEstoy pensando en un número entre 1 y 100.")
  print(f"Tienes {max_intentos} intentos. ¡Buena suerte!\n")

  # Usamos ciclo while
  while intentos < max_intentos:
      entrada = input(f"Intento {intentos + 1}/{max_intentos} — Adivina el número (1-100): ").strip().lower()

      # Validación 1: ¿El usuario quiere salir?
      if entrada == "salir":
        print("Saliste del juego.")
        break

      # Validación 2: ¿es número?
      if not entrada.isnumeric():
        print("Eso no es un número. Ingresa un número.")
        continue

      # Convertir a entero
      num_adivina = int(entrada)
      intentos += 1

      # Logica del juego
      if 1 <= num_adivina <= 100:
        if num_adivina > num_secret:
            print("Muy alto")
        elif num_adivina < num_secret:
            print("Muy bajo")
        else:
            print(f"¡Correcto! Lo lograste en {intentos} intentos")
            break
      else:
        print("El numero debe estar en un rango de 1 a 100")
      print("-" * 40)

  else:
    print(f"Se acabaron los intentos. El número era {num_secret}.")


  print("_" * 40)
  new_game = input("¿Quieres jugar de nuevo? (si/no): ").strip().lower()
  if new_game == "no":
    print("\n¡Gracias por jugar! Hasta la próxima.")
    break
```

```python
###   RETO A - JUEGO ADIVINA EL NÚMERO   ###

import random

print("=" * 40)
print("        JUEGO ADIVINA EL NÚMERO")
print("=" * 40)


while True:
  # Validación 0: ¿Qué nivel quieres jugar?
  nivel_juego = input("¿Qué nivel quieres jugar? (facil, normal, dificil): ").strip().lower()

  if nivel_juego == "facil":
      num_secret = random.randint(1, 50)
      max_intentos = 10
      rango_max = 50
  elif nivel_juego == "normal":
      num_secret = random.randint(1, 100)
      max_intentos = 7
      rango_max = 100
  elif nivel_juego == "dificil":
      num_secret = random.randint(1, 200)
      max_intentos = 6
      rango_max = 200
  else:
    print("El nivel escogido no es correcto. Inténtalo de nuevo. ")    
    

  intentos = 0
  print(f"\nEstoy pensando en un número entre 1 y {rango_max}.")
  print(f"Tienes {max_intentos} intentos. ¡Buena suerte!\n")

  # Usamos ciclo while
  while intentos < max_intentos:
      entrada = input(f"Intento {intentos + 1}/{max_intentos} — Adivina el número (1-{rango_max}): ").strip().lower()

      # Validación 1: ¿El usuario quiere salir?
      if entrada == "salir":
        print("Saliste del juego.")
        break

      # Validación 2: ¿es número?
      if not entrada.isnumeric():
        print("Eso no es un número. Ingresa un número.")
        continue

      # Convertir a entero
      num_adivina = int(entrada)
      intentos += 1

      # Logica del juego
      if 1 <= num_adivina <= rango_max:
        if num_adivina > num_secret:
            print("Muy alto")
        elif num_adivina < num_secret:
            print("Muy bajo")
        else:
            print(f"¡Correcto! Lo lograste en {intentos} intentos")
            break
      else:
        print(f"El numero debe estar en un rango de 1 a {rango_max}")
      print("-" * 40)

  else:
    print(f"Se acabaron los intentos. El número era {num_secret}.")


  print("_" * 40)
  new_game = input("¿Quieres jugar de nuevo? (si/no): ").strip().lower()
  if new_game == "no":
    print("\n¡Gracias por jugar! Hasta la próxima.")
    break
```
